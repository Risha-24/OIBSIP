```python
df['Date'].head()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date'].dtype
```

```python
monthly_sales = df.groupby(
    df['Date'].dt.to_period('M')
)['Total_Amount'].sum()

monthly_sales
```

```python
plt.figure(figsize=(12, 5))

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
quarterly_sales = df.groupby(
    df['Date'].dt.to_period('Q')
)['Total_Amount'].sum()

quarterly_sales
```

```python
plt.figure(figsize=(10, 5))

quarterly_sales.plot(kind='line', marker='o')

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 18, 25, 35, 45, 55, 65, 100],
    labels=['<18', '18-25', '26-35', '36-45', '46-55', '56-65', '65+']
)

df['Age_Group'].value_counts().sort_index()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='Age_Group')

plt.title('Customer Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)

plt.show()
```

```python
df['Gender'].value_counts()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Gender')

plt.title('Customer Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')

plt.show()
```

### Observation

- The age group distribution shows which customer age group has the highest number of customers.
- The gender distribution shows the proportion of male and female customers.
- These demographic patterns can help businesses understand their major customer segments and design targeted marketing strategies.

```python
category_revenue = df.groupby('Product_Category')['Total_Amount'].sum().sort_values(ascending=False)

print(category_revenue)
```

```python
plt.figure(figsize=(10, 5))

category_revenue.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

### Observation

- The correlation heatmap shows the relationships between the numerical variables.
- Strong positive or negative correlations can indicate meaningful relationships between variables.
- The relationship between Quantity, Unit Price and Total Amount is particularly useful for understanding sales behaviour.
- Correlation analysis helps identify variables that may influence customer purchases and revenue.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Discount_Amount',
    y='Total_Amount'
)

plt.title('Discount Amount vs Total Sales Amount')
plt.xlabel('Discount Amount')
plt.ylabel('Total Amount')

plt.show()
```

### Observation: Discount Amount vs Total Sales Amount

The scatter plot shows the relationship between discount amount and total sales amount. The distribution of points helps us understand how sales amounts vary across different discount values. Further analysis may be required to determine whether discounts influence sales performance.
